## Задача 1. Оптимальна ціна товару
**Умова**
Ви працюєте маркетологом в онлайн-магазині електроніки.

За результатами досліджень, попит на навушники залежить від ціни за формулою:
$Q(x) = 2000 - 0.8x$

Собівартість одних навушників становить 800 грн. Тоді прибуток компанії:
$P(x) = (x - 800) \cdot Q(x) = (x - 800)(2000 - 0.8x)$

## Задача 2. Мінімізація функції двох змінних
**Умова**
Розглянемо функцію:
$f(x, y) = x^2 + xy + y^2 - 6x - 9y + 20$
Потрібно знайти точку мінімуму аналітично та перевірити результат чисельними методами.

**Завдання**
1. **Знайдіть критичну точку аналітично.** Обчисліть частинні похідні $\frac{\partial f}{\partial x}$ та $\frac{\partial f}{\partial y}$. Прирівняйте їх до нуля та розв'яжіть систему рівнянь, щоб знайти критичну точку $(x^*, y^*)$.
2. **Перевірте тип екстремуму.** Далі аналітично обчисліть другі частинні похідні та складіть матрицю Гессе. Знайдіть власні значення матриці через характеристичне рівняння $\det(H - \lambda I) = 0$. Згадуємо власні значення з Теми 3. Робимо висновок — знайдена точка є мінімумом або максимумом чи взагалі сідловою точкою.
3. **Знайдіть оптимум чисельно.** Напишіть програму, яка використовує функцію `minimize` з бібліотеки `scipy.optimize` з методом `BFGS`. Запустіть оптимізацію з початкової точки (0, 0).
4. **Перевірте стійкість розв'язку.** Запустіть програмну оптимізацію з трьох різних початкових точок: (0, 0), (10, 10), (-5, 15). Переконайтесь, що всі запуски збігаються до однієї точки мінімуму.


```python
# Твоє рішення для Задачі 2
from scipy.optimize import minimize
import numpy as np

#df_x = 2*x+y-6
#df_y = x+2*y-9

#2*x+y-6 = 0
#x+2*y-9 = 0

#x = 1
#y = 4

critical_point = np.array([1, 4])

#df_xx = 2
#df_yy = 2
#df_xy = 1
#df_yx = 1

matrix_gs = np.array([
    [2, 1],
    [1, 2]
])

eigh = np.linalg.eigvals(matrix_gs)

print(f'Критичні точки {critical_point}')
print(f' Матриця Гассе {matrix_gs}')
print(f'Власні значення {eigh}')

if np.all(eigh>0):
  print("Точка мінімальна")
elif np.all(eigh<0):
  print(' Точка максимальна')
else:
  print('Сідлова точка')
```

```python
def f(vars):
  x,y = vars
  return  x**2+x*y+y**2-6*x-9*y+20

initial_points = [
    [0,0],
    [10,10],
    [-5,15]
]

for x,y in enumerate(initial_points, 1):
  res = minimize(f, y, method="BFGS")
  faund_x, faund_y = res.x
  print(f'Запуск {x} ітерації, старт з точки {y}')
  print(f'Знайдена точка x = {faund_x:.4f}, y = {faund_y:.4f}')
  print(f'Значення функції в мінімумі {res.fun:.4f}')
```

## Задача 3. Оптимальний план виробництва
**Умова**
Меблева фабрика виготовляє два види продукції: стільці та столи. Для виробництва використовуються два ресурси: деревина та робочий час.

| Ресурс | Витрати на 1 стілець | Витрати на 1 стіл | Загальний запас |
| :--- | :--- | :--- | :--- |
| Деревина (м²) | 2 | 4 | 120 |
| Робочий час (год) | 3 | 2 | 90 |

Прибуток від продажу одного стільця — 500 грн, одного столу — 800 грн.
Потрібно знайти оптимальний план виробництва, який максимізує прибуток.

**Завдання**
1. **Сформулюйте задачу лінійного програмування аналітично.** Введіть змінні $x$ (кількість стільців) та $y$ (кількість столів). Запишіть цільову функцію прибутку $P(x, y)$, яку потрібно максимізувати. Запишіть систему обмежень на ресурси та умови невід'ємності.
2. **Розв'яжіть задачу програмно.** Використайте функцію `linprog` з бібліотеки `scipy.optimize`. Пам'ятайте, що `linprog` виконує мінімізацію, тому для максимізації прибутку потрібно мінімізувати $-P(x, y)$.
3. **Проаналізуйте використання ресурсів.** Далі в програмі обчисліть, скільки кожного ресурсу буде використано при оптимальному плані. Визначте, які ресурси використані повністю, а які мають залишок.


```python
# Твоє рішення для Задачі 3
from scipy.optimize import linprog

# p(x,y) = 500x+800y
# 2x+4y<=120
# 3x+2y<=90
# x, y >= 0

c = [-500, -800]
A = [
    [2, 4],
    [3, 2]
]
B = [120, 90]

x_lim = (0, None)
y_lim = (0, None)

```

```python
result = linprog(c, A, B, bounds=[x_lim, y_lim], method='highs')
optimal_x, optimal_y = result.x
max_profit = -result.fun
print(f'Оптимальна клькість стільців {optimal_x}, столів {optimal_y}, максимальний прибуток {max_profit}')
```

```python
used_wood = optimal_x*2+optimal_y*4
used_time = optimal_x*3+optimal_y*2
print(f'Використано дерева {used_wood} залишок дерева {120-used_wood}')
print(f'Використано часу {used_time}, залишок часу {90-used_time}')
```

## Задача 4. Комплексний аналіз виробництва
**Умова**
Це фіналізуюча задача дисципліни, яка поєднує інструменти з різних тем: диференціювання, інтегрування, системи рівнянь та оптимізацію.
Завод виробляє продукцію протягом 10-годинної зміни.

Продуктивність (одиниць на годину) змінюється протягом дня за формулою:
$P(t) = 100 + 40t - 4t^2$
де $t$ — час у годинах від початку зміни $0 \le t \le 10$.

Вартість виробництва однієї одиниці залежить від двох параметрів: витрат на сировину $x$ та витрат на енергію $y$ (грн):
$C(x, y) = x^2 + y^2 - 10x - 8y + 50$

Початкові значення параметрів визначаються з бюджетних обмежень:
$\begin{cases} 2x + y = 20 \\ x + 3y = 25 \end{cases}$

**Завдання**
Напишіть програму де виконайте всі зазначені нижче пункти.

**Частина перша. Дослідження продуктивності**
1. **Проаналізуйте динаміку продуктивності.** Використайте `approx_fprime` для обчислення швидкості зміни продуктивності $P'(t)$ у три моменти: $t = 2, t = 5, t = 8$. Визначте, коли продуктивність зростає, а коли спадає.
2. **Знайдіть момент пікової продуктивності.** Використайте `minimize_scalar` для пошуку моменту $t^*$, коли продуктивність максимальна. Обчисліть значення $P(t^*)$.
3. **Обчисліть загальний обсяг виробництва.** Використайте `scipy.integrate.quad` для обчислення інтегралу $\int_0^{10} P(t) dt$ — загальної кількості одиниць за зміну.

**Частина друга. Оптимізація витрат**
1. **Визначте початкові параметри.** Використайте `numpy.linalg.solve` для розв'язання системи рівнянь та знаходження початкових значень $(x_0, y_0)$.
2. **Мінімізуйте вартість виробництва.** Використайте `scipy.optimize.minimize` з методом `BFGS` для пошуку оптимальних параметрів $(x^*, y^*)$, що мінімізують $C(x, y)$. Початкова точка — результат з пункту 4.

**Фінал**
1. **Обчисліть загальну вартість.** Помножте загальний обсяг виробництва (пункт 3) на мінімальну вартість одиниці (пункт 5).


```python
# Твоє рішення для Задачі 4
from scipy.optimize import approx_fprime, minimize_scalar, minimize
from scipy.integrate import quad

time_point = [2.0, 5.0, 8.0]

def P(t):
  return 100+40*t-4*t**2


for t in time_point:
  speed_of_change = approx_fprime(np.array([t]), P)
  if speed_of_change > 0:
    print("Зростає продуктивність")
  else:
    print("Спадає продуктивність")
  print(f'Переіряємо t={t}, i швидкість зміни {speed_of_change[0]:.6f} одиниць на годину квадраних')


```

```python
def P_negative(t):
  return -(100+40*t-4*t**2)

rs_pik = minimize_scalar(P_negative, bounds = (0, 10), method= "bounded")
pik_time = rs_pik.x
pik_value = -rs_pik.fun

print(f'Момент пікової продуктивності  t*={pik_time:.2f}')
print(f'Максимальна продуктивність p(t*)={pik_value:.2f}')
```

```python
total_volume, error = quad(P, 0, 10)
print(f'Загальний обсяг виробництва за зміну {total_volume:.2f} ')
```

```python
def C(vars):
  x,y = vars
  return x**2+y**2-10*x-8*y+50
A = np.array([
    [2, 1],
    [1, 3]
])

b = np.array([20, 25])

start_point = np.linalg.solve(A, b)
print(f'Початкова точка параметрів (x0, y0) {start_point}')
```

```python
rs_cost = minimize(C, start_point, method= 'BFGS')
optimal_x, optimal_y = rs_cost.x
min_cost = rs_cost.fun
print(f'Оптимальні параметри (x*,y*), x={optimal_x:.2f}, y={optimal_y:.2f}')
print(f'Мінімальна вартість однієі одиниці {min_cost:.2f}')
```
