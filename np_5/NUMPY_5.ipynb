## Задача 1. Аналіз рейтингів мобільного додатку
**Умова**
Ви аналізуєте відгуки на мобільний додаток в App Store. За останній місяць користувачі залишили 2000 оцінок:

| Оцінка | Кількість відгуків |
| :--- | :--- |
| 1 ⭐ | 120 |
| 2 ⭐ | 180 |
| 3 ⭐ | 320 |
| 4 ⭐ | 580 |
| 5 ⭐ | 800 |

Випадкова величина $X$ — оцінка, яку ставить випадковий користувач.

**Завдання**
1. Побудуйте функцію маси ймовірності PMF. Для кожної оцінки обчисли ймовірність $P(X = x) = \frac{\text{кількість відгуків з оцінкою } x}{\text{загальна кількість відгуків}}$
2. Побудуйте функцію розподілу CDF. Для кожної оцінки обчисли накопичену ймовірність $F(x) = P(X \le x)$. Підказка: Оскільки CDF — це накопичена сума ймовірностей, використайте `np.cumsum()`.
3. Виведіть результат у вигляді таблиці з трьома колонками: оцінка, PMF, CDF.
4. Обчисліть математичне сподівання $E[X] = \sum x \cdot P(X = x)$, дисперсію $Var(X) = E[X^2] - (E[X])^2$ та стандартне відхилення $\sigma = \sqrt{Var(X)}$.
5. Яка ймовірність, що випадковий користувач поставить негативну оцінку (1 або 2 зірки)?
6. Яка ймовірність, що користувач поставить 4 або більше зірок?
7. Знайдіть медіану рейтингу — найменше значення $x$, для якого $F(x) \ge 0.5$


```python
# Твоє рішення для Задачі 1
import numpy as np

reitings_vector = np.array([1, 2, 3, 4, 5])
counts_vector = np.array([120, 180, 320, 580, 800])
total_revu = np.sum(counts_vector)
PMF = counts_vector/total_revu
CDF = np.cumsum(PMF)

print('Оцінка  | PMF    | CDF    |')
print('-'*27)
for i,j,k in zip(reitings_vector, PMF, CDF):
  print(f'{i:<7}| {j:.4f} | {k:.4f} |')

# Завдання 4: Математичне сподівання, дисперсія та стандартне відхилення
EX = np.sum(reitings_vector * PMF)
EX2 = np.sum((reitings_vector**2) * PMF)
VarX = EX2 - (EX**2)
SigmaX = np.sqrt(VarX)

print(f'\nМатематичне сподівання E[X]: {EX:.2f}')
print(f'Дисперсія Var(X): {VarX:.2f}')
print(f'Стандартне відхилення sigma: {SigmaX:.2f}')

# Завдання 5: Ймовірність негативної оцінки (1 або 2 зірки)
p_negative = PMF[0] + PMF[1] # P(X=1) + P(X=2)
# Або P(X <= 2) = CDF[1]
print(f'\nЙмовірність негативної оцінки (1 або 2 зірки): {p_negative:.4f}')

# Завдання 6: Ймовірність 4 або більше зірок
p_4_or_more = PMF[3] + PMF[4] # P(X=4) + P(X=5)
# Або P(X >= 4) = 1 - CDF[2]
print(f'Ймовірність 4 або більше зірок: {p_4_or_more:.4f}')

# Завдання 7: Медіана рейтингу
median_index = np.where(CDF >= 0.5)[0][0]
median_rating = reitings_vector[median_index]
print(f'Медіана рейтингу: {median_rating}')
```

## Задача 2. Аналіз доставки замовлень
**Умова**
Ви аналізуєте роботу служби доставки інтернет-магазину. За статистикою, кожне замовлення доставляється вчасно з ймовірністю 0.88. Сьогодні відправлено 50 замовлень.

Кількість вчасно доставлених замовлень $X$ описується біноміальним розподілом $X \sim Binomial(n = 50, p = 0.88)$.

**Завдання**
1. Обчисліть математичне сподівання $E[X]$ та стандартне відхилення $\sigma$ кількості вчасно доставлених замовлень.
2. Яка ймовірність, що всі 50 замовлень доставлять вчасно?
3. Яка ймовірність, що рівно 45 замовлень доставлять вчасно?
4. Яка ймовірність, що вчасно доставлять від 42 до 46 замовлень?
5. Яка ймовірність, що більше 5 замовлень запізняться?
Використайте `scipy.stats.binom` для обчислень.


```python
# Твоє рішення для Задачі 2
from scipy.stats import binom

n = 50
p = 0.88

# Завдання 1: Математичне сподівання та стандартне відхилення
mean_X = binom.mean(n, p)
std_X = binom.std(n, p)
print(f'Математичне сподівання E[X]: {mean_X:.2f}')
print(f'Стандартне відхилення sigma: {std_X:.2f}')

# Завдання 2: Ймовірність, що всі 50 замовлень доставлять вчасно
p_all_50_ontime = binom.pmf(50, n, p)
print(f'\nЙмовірність, що всі 50 замовлень доставлять вчасно: {p_all_50_ontime:.6f}')

# Завдання 3: Ймовірність, що рівно 45 замовлень доставлять вчасно
p_exact_45_ontime = binom.pmf(45, n, p)
print(f'Ймовірність, що рівно 45 замовлень доставлять вчасно: {p_exact_45_ontime:.6f}')

# Завдання 4: Ймовірність, що вчасно доставлять від 42 до 46 замовлень
p_42_to_46_ontime = binom.cdf(46, n, p) - binom.cdf(41, n, p)
print(f'Ймовірність, що вчасно доставлять від 42 до 46 замовлень: {p_42_to_46_ontime:.6f}')

# Завдання 5: Ймовірність, що більше 5 замовлень запізняться
# Це означає, що менше ніж 45 замовлень будуть вчасно (50 - 5 = 45)
# P(late > 5) = P(ontime < 45) = P(ontime <= 44)
p_more_than_5_late = binom.cdf(44, n, p)
print(f'Ймовірність, що більше 5 замовлень запізняться: {p_more_than_5_late:.6f}')
```

## Задача 3. Час між замовленнями в інтернет-магазині
**Умова**
Ви аналізуєте потік замовлень в інтернет-магазині. В середньому надходить 6 замовлень на годину, тобто одне замовлення кожні 10 хвилин.

Час між замовленнями $T$ описується показниковим розподілом з параметром $\lambda = 6$ замовлень на годину або $\lambda = 0.1$ замовлень на хвилину.

Пам'ятаємо, що для показникового розподілу в `scipy.stats.expon` параметр `scale` — це середній час між подіями: $scale = 1/\lambda = 10$ хвилин.

**Завдання**
1. Обчисліть математичне сподівання $E[T]$ та стандартне відхилення $\sigma$ часу між замовленнями.
2. Яка ймовірність, що наступне замовлення надійде менше ніж за 5 хвилин?
3. Яка ймовірність, що доведеться чекати більше 15 хвилин?
4. Яка ймовірність, що замовлення надійде в проміжку від 5 до 15 хвилин?
5. Знайдіть медіанний час очікування — час, за який з ймовірністю 50% надійде наступне замовлення.
Використайте `scipy.stats.expon` для обчислень.


```python
# Твоє рішення для Задачі 3
from scipy.stats import expon
import numpy as np

# Параметр lambda = 0.1 замовлень на хвилину, scale = 1/lambda = 10 хвилин
scale_minutes = 10

# Завдання 1: Математичне сподівання та стандартне відхилення
mean_T = expon.mean(scale=scale_minutes)
std_T = expon.std(scale=scale_minutes)
print(f'Математичне сподівання E[T]: {mean_T:.2f} хвилин')
print(f'Стандартне відхилення sigma: {std_T:.2f} хвилин')

# Завдання 2: Ймовірність, що наступне замовлення надійде менше ніж за 5 хвилин
p_less_than_5_min = expon.cdf(5, scale=scale_minutes)
print(f'\nЙмовірність, що наступне замовлення надійде менше ніж за 5 хвилин: {p_less_than_5_min:.4f}')

# Завдання 3: Ймовірність, що доведеться чекати більше 15 хвилин
p_more_than_15_min = 1 - expon.cdf(15, scale=scale_minutes)
print(f'Ймовірність, що доведеться чекати більше 15 хвилин: {p_more_than_15_min:.4f}')

# Завдання 4: Ймовірність, що замовлення надійде в проміжку від 5 до 15 хвилин
p_5_to_15_min = expon.cdf(15, scale=scale_minutes) - expon.cdf(5, scale=scale_minutes)
print(f'Ймовірність, що замовлення надійде в проміжку від 5 до 15 хвилин: {p_5_to_15_min:.4f}')

# Завдання 5: Медіанний час очікування
median_T = expon.ppf(0.5, scale=scale_minutes)
print(f'Медіанний час очікування: {median_T:.2f} хвилин')
```

## Задача 4. Перевірка середнього балу студентів
**Умова**
Минулого року середній бал студентів на іспиті з математики становив 72 бали. Цього року викладач змінив методику підготовки і хоче перевірити, чи вплинуло це на результати.

Зібрано результати 35 випадкових студентів:
* Вибіркове середнє: $\bar{x} = 76.8$ балів
* Вибіркове стандартне відхилення: $s = 14.2$ балів
* Розмір вибірки: $n = 35$

**Завдання**
1. Сформулюйте нульову гіпотезу $H_0$ та альтернативну гіпотезу $H_1$.
2. Обчисліть стандартну похибку середнього $SE = \frac{s}{\sqrt{n}}$
3. Обчисліть t-статистику $t = \frac{\bar{x} - \mu_0}{SE}$, де $\mu_0 = 72$.
4. Знайдіть p-value для двостороннього тесту, використовуючи `scipy.stats.t.cdf`.
5. Побудуйте 95% довірчий інтервал для середнього балу. Використовуйте `scipy.stats.t.ppf` для знаходження критичного значення $t_{crit}$
6. Зробіть висновок при рівні значущості $\alpha = 0.05$: чи є підстави стверджувати, що нова методика вплинула на результати?


```python
# Твоє рішення для Задачі 4
from scipy.stats import t
import numpy as np

# Дані
mean_sample = 76.8 # Вибіркове середнє
std_sample = 14.2  # Вибіркове стандартне відхилення
n = 35             # Розмір вибірки
mu_0 = 72          # Середній бал минулого року (значення для H0)
alpha = 0.05       # Рівень значущості

# Завдання 1: Нульова та альтернативна гіпотези
print(f'1. Нульова гіпотеза H0: Середній бал студентів не змінився (mu = {mu_0})')
print(f'   Альтернативна гіпотеза H1: Середній бал студентів змінився (mu != {mu_0})')

# Завдання 2: Стандартна похибка середнього
SE = std_sample / np.sqrt(n)
print(f'\n2. Стандартна похибка середнього (SE): {SE:.3f}')

# Завдання 3: t-статистика
t_statistic = (mean_sample - mu_0) / SE
print(f'3. t-статистика: {t_statistic:.3f}')

# Завдання 4: p-value для двостороннього тесту
df = n - 1 # Ступені свободи
p_value = 2 * (1 - t.cdf(abs(t_statistic), df))
print(f'4. p-value: {p_value:.4f}')

# Завдання 5: 95% довірчий інтервал
t_critical = t.ppf(1 - alpha/2, df)
margin_of_error = t_critical * SE
conf_interval_lower = mean_sample - margin_of_error
conf_interval_upper = mean_sample + margin_of_error
print(f'5. 95% Довірчий інтервал: ({conf_interval_lower:.2f}, {conf_interval_upper:.2f})')

# Завдання 6: Висновок
print(f'\n6. Висновок при рівні значущості alpha = {alpha}:')
if p_value < alpha:
    print(f'   Оскільки p-value ({p_value:.4f}) < alpha ({alpha}), ми відкидаємо нульову гіпотезу.')
    print(f'   Є достатньо підстав стверджувати, що нова методика вплинула на середній бал студентів.')
else:
    print(f'   Оскільки p-value ({p_value:.4f}) >= alpha ({alpha}), ми НЕ відкидаємо нульову гіпотезу.')
    print(f'   Немає достатньо підстав стверджувати, що нова методика вплинула на середній бал студентів.')
```
