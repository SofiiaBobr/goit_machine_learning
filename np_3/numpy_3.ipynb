## Задача 1. Аналітика помилок у логах
**Умова**
Ви працюєте DevOps-інженером і аналізуєте логи системи моніторингу. За добу зафіксовано 500 подій. З них:
* 85 — помилки бази даних з тегом `DB`
* 60 — помилки мережі з тегом `NET`
* 35 — помилки, що мають обидва теги одночасно `DB` та `NET`

Позначимо:
* $\Omega$ — множина всіх 500 подій
* $A$ — подія "помилка бази даних" з тегом `DB`
* $B$ — подія "помилка мережі" з тегом `NET`

**Завдання:**
1. Яка ймовірність, що випадково вибрана подія — помилка DB?
2. Яка ймовірність, що подія має тег DB **або** NET?
3. Яка ймовірність, що подія має тег DB, але **не** NET?


```python
# Твоє рішення для Задачі 1
omega = 500
a = 85
b = 60
a_and_b = 35

pro_db = a/omega
print(f'Ймовірність помилки DB {pro_db*100:.2f}%')
```

```python
db_or_net = a+b-a_and_b
pro_db_or_net = db_or_net/omega
print(f'Ймовірність помилки DB або NET {pro_db_or_net*100:.2f}%')
```

```python
pro_db_not_net = (a-a_and_b)/omega
print(f'Ймовірність помилкм DB але не NET {pro_db_not_net*100:.2f}%')
```

## Задача 4. Визначення джерела трафіку
**Умова**
Інтернет-магазин отримує трафік з трьох джерел:
* Пошукова реклама: 50% трафіку
* Соцмережі: 30% трафіку
* Email-розсилка: 20% трафіку

Конверсія (ймовірність покупки) для кожного джерела:
* Пошукова реклама: 4%
* Соцмережі: 2%
* Email-розсилка: 8%

Позначимо:
* $H_1$ — подія "відвідувач прийшов з пошукової реклами"
* $H_2$ — подія "відвідувач прийшов з соцмереж"
* $H_3$ — подія "відвідувач прийшов з email-розсилки"
* $A$ — подія "відвідувач зробив покупку"

**Завдання**
1. Яка загальна ймовірність, що випадковий відвідувач зробить покупку?
2. Відвідувач зробив покупку. Яка ймовірність, що він прийшов з email-розсилки?
3. Напишіть функцію `bayes_source(prior_probs, conversion_rates)`, яка для довільної кількості джерел обчислює апостеріорні ймовірності кожного джерела після здійснення покупки.
Параметри функції:
* `prior_probs` — список апріорних ймовірностей кожного джерела
* `conversion_rates` — список конверсій для кожного джерела
4. Функція повертає список апостеріорних ймовірностей для кожного джерела.


```python
# Твоє рішення для Задачі 4
P_h1 = 0.5
P_h2 = 0.3
P_h3 = 0.2

P_A_h1 = 0.04
P_A_h2 = 0.02
P_A_h3 = 0.08

P_A = P_h1*P_A_h1+P_h2*P_A_h2+P_h3*P_A_h3
print(f' Загальна вірогідність покупки {P_A*100:.2f}%')




```

```python
P_h3_given_A = (P_h3*P_A_h3)/P_A
print(f' Ймовірність що покупець прийшов з емейл {P_h3_given_A*100:.2f}%')
```

```python
def bayes_source(prior_probs, conversion_rates):
    if len(prior_probs) != len(conversion_rates):
      raise ValueError('Списки мають співпадати')
    total_prob = 0
    for i,j in zip(prior_probs, conversion_rates):
      total_prob += i*j
    posterior_probs = [(i*j)/total_prob for i,j in zip(prior_probs, conversion_rates)]


    source_names = ["Пошукова реклама", "Соцмережі", "Email-розсилка"]
    print("Апостеріорні ймовірності після покупки:")
    for i, prob in enumerate(posterior_probs):
        if i < len(source_names):
            print(f"- {source_names[i]}: {prob*100:.2f}%")
        else:
            print(f"- Джерело {i+1}: {prob*100:.2f}%")

    return posterior_probs

prior_probs = [0.5, 0.3, 0.2]
conversion_rates = [0.04, 0.02, 0.08]
```

```python
print(bayes_source(prior_probs, conversion_rates))
```
