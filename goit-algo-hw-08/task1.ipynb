```python
class Node:
    def __init__(self, key):
        self.left = None
        self.right = None
        self.val = key



# Insert into BST


def insert(root, key):

    if root is None:
        return Node(key)

    if key < root.val:
        root.left = insert(root.left, key)

    else:
        root.right = insert(root.right, key)

    return root



# Task 1
# Find minimum value


def find_min(root):

    current = root

    while current.left is not None:
        current = current.left

    return current.val



# Task 2
# Sum all values


def find_sum(root):

    if root is None:
        return 0

    return (
        root.val
        + find_sum(root.left)
        + find_sum(root.right)
    )



# Task 3
# Minimize cable connection cost


import heapq


def min_cable_cost(cables):

    heapq.heapify(cables)

    total_cost = 0

    while len(cables) > 1:

        first = heapq.heappop(cables)
        second = heapq.heappop(cables)

        cost = first + second

        total_cost += cost

        heapq.heappush(cables, cost)

    return total_cost



# Main


def main():

    root = None

    values = [10, 5, 3, 7, 15, 12, 18]

    for value in values:
        root = insert(root, value)

    print("Task 1")
    print("Minimum value in BST:", find_min(root))

    print("\nTask 2")
    print("Sum of all values:", find_sum(root))

    print("\nTask 3")

    cables = [4, 3, 2, 6]

    print(
        "Minimum cable connection cost:",
        min_cable_cost(cables)
    )


if __name__ == "__main__":
    main()
```
