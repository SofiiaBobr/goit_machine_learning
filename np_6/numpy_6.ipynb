## Завдання 1
Потрібно розробити програму, яка імітує приймання й обробку заявок: програма має автоматично генерувати нові заявки (ідентифіковані унікальним номером або іншими даними), додавати їх до черги, а потім послідовно видаляти з черги для "обробки", імітуючи таким чином роботу сервісного центру.

У комірці з кодом нижче наведено псевдокод для створення системи обробки заявок. У ньому використовуються дві основні функції: `generate_request()`, яка генерує нові заявки та додає їх до черги, та `process_request()`, яка обробляє заявки, видаляючи їх із черги. Головний цикл програми виконує ці функції, імітуючи постійний потік нових заявок та їх обробку.


```python
# Твоє рішення для Завдання 1
from queue import Queue
import time
import random

queue = Queue()
request_counter = 0

def generate_request():
  global request_counter
  request_counter +=1
  request_id = f'req-{request_counter}'
  print(f'Створено нову заявку id-{request_id}')
  queue.put(request_id)


def process_request():
  if not queue.empty():
    current_request = queue.get()
    print(f'Заявку {current_request} успішно опрацьовано')
  else:
    print(f'Черга порожня')

for i in range(1, 11):
  print(f'Крок симуляції {i}')
  time.sleep(1)
  if random.choice([True, True, True, False]):
    generate_request()
  if random.choice([True, True, False]):
    process_request()




```

## Завдання 2
Необхідно розробити функцію, яка приймає рядок як вхідний параметр, додає всі його символи до двосторонньої черги (`deque` з модуля `collections` в Python), а потім порівнює символи з обох кінців черги, щоб визначити, чи є рядок паліндромом. Програма повинна правильно враховувати як рядки з парною, так і з непарною кількістю символів, а також бути нечутливою до регістру та пробілів.


```python
# Твоє рішення для Завдання 2
from collections import deque

def is_palindrome(text):
    text_lower = text.lower()
    clear_text = "".join(text_lower.split())
    char_deque = deque(clear_text)
    while len(char_deque)>1:
      left_char = char_deque.popleft()
      right_char = char_deque.pop()
      if left_char != right_char:
         return False
    return True



# Тестування
print(is_palindrome("А роза упала на лапу Азора")) # Має повернути True
```
